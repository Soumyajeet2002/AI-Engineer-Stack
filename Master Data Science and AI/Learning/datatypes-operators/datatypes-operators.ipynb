# Typecasting
- what do we mean by TYPECASTING 
    - typecasting is done when we need to change the data type for value/data.

```python
age = 25.5
marks = 75.6

new_age = int(age)
final_marks = int(marks)

print("Original age: " + str(age))
print("Original marks: " + str(marks))
print("After TypeCasting to INT: " + str(new_age))
print("After TypeCasting to INT: " + str(final_marks))
```

# In typecasting we will loose some Values
- like from float 9.5 if we convert it to INT it will give us 9 -> in this we have loast 0.5 value.

```python
var = 2
print(type(var))

var_new = float(var)
print("value after typecasting: " + str(var_new))
print(type(var_new))
```

# String
##
##
## String is ImMutable in nature 
- this means string can not be edited , but STRING IS INDEXED WISE.
- That means once a string is created, you cannot change individual characters in it.
- But we can edit string by creating new STRING.

```python
val = "HELLO"

print(val[0])
print(val[1])
print(val[2])
print(val[3])
print(val[4])


val = "k" + val[1:]
print(val)
```

```python
s = "Hello"
print(s)
s = "J" + s[1:]
print(s)
```
