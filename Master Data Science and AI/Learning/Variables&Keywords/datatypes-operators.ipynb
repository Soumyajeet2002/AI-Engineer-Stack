# Data Types
## What is Data Types
- data types are the types of data we have.
- NUMERIC
- DICTIONARY
- BOOLEAN
- SET
- SEQUENCE TYPE

```python
# Finding the Data Types this type() tells the data types assigned to variables
x = 5
y = 10.5
z = "Hello, World!"

print(x)
print(y)
print(z)

print(type(x))
print(type(y))
print(type(z))
```

```python
name = input("Enter you name: ")
print(name)

print(type(name))
```

# Typecasting
- what do we mean by TYPECASTING 
    - typecasting is done when we need to change the data type for value/data.

```python
age = 25.5 
marks = 75.6

new_age = int(age)
final_marks = int(marks)

print("Original age: "  + str(age))
print("Original marks: " + str(marks))
print("After TypeCasting to INT: " + str(new_age))
print("After TypeCasting to INT: " + str(final_marks))
```
