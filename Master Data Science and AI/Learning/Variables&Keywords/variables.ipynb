# Creating a SUM function

```python
a = 50
b = 100


def sum_function(a, b):
    c = a + b
    print(c)
    return "sum function executed successfully"


sum_function(a, b)
```

```python
c = 100
d = 300


def sum_function(c, d):
    e = c + d
    f = c - d
    g = c * d
    h = c / d
    print(e)
    print(f)
    print(g)
    print(h)
    return "sum function executed successfully"


sum_function(c, d)
```
